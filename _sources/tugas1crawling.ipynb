Kode ini saya buat untuk melakukan crawling artikel dari Springer menggunakan **RSS feed** berdasarkan keyword tertentu (web mining, web usage mining, fraud detection). Hasil crawling berupa **judul, link, ringkasan, dan tanggal** publikasi artikel kemudian dikumpulkan dan disimpan ke dalam file CSV agar mudah dianalisis lebih lanjut.

## code utama

- Mencari artikel di Springer berdasarkan beberapa keyword.
- Mengambil data artikel dari RSS feed (judul, link, ringkasan, tanggal publikasi).
- Mengumpulkan semua hasil ke dalam satu DataFrame pandas.
- Menyimpan hasil crawling ke file CSV.

```python
# Daftar keyword
keywords = ["web mining", "web usage mining", "fraud detection"]
all_results = []

for kw in keywords:
    url = f"https://link.springer.com/search.rss?query={kw.replace(' ', '+')}"
    print(f"[INFO] Ambil RSS untuk: {kw}")
    feed = feedparser.parse(url)
    
    for entry in feed.entries:
        all_results.append({
            "keyword": kw,
            "title": entry.title,
            "link": entry.link,
            "summary": entry.summary if "summary" in entry else "No summary",
            "pubDate": entry.get("published", None)
        })

# Simpan ke CSV
df = pd.DataFrame(all_results)
df.to_csv("springer_rss_results.csv", index=False, encoding="utf-8")

print("[INFO] Data berhasil disimpan ke springer_rss_results.csv")
df.head(10)  # preview 10 baris pertama dengan tampilan tabel pandas
```
