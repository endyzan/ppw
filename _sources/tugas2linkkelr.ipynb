# Tugas 2 | Link Keluaran

Tugas 2 link keluaran ini tujuannya untuk merekam riwayat navigasi
Saat user membuka browser dan menjelajahi situs,
setiap perubahan URL akan direkam (dari link asal
ke link tujuan), lalu hasilnya disimpan ke CSV.

Berikut Code nya:

**Import Library**

```python
import sys
!{sys.executable} -m pip install selenium
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
import csv
import time
```

```python
# Konfigurasi Chrome
chrome_options = Options()

# Inisialisasi browser Chrome dengan opsi yang sudah ditentukan
driver = webdriver.Chrome(options=chrome_options)

# Buka website awal (dalam contoh: Fakultas Hukum UTM)
driver.get("https://hukum.trunojoyo.ac.id/")

# List untuk menyimpan catatan riwayat URL
records = []

# Simpan URL awal
last_url = driver.current_url

print("🔎 Browser terbuka. Silakan jelajahi situs secara manual.")
print("⏳ Tekan CTRL + C di terminal jika sudah selesai.\n")

try:
    while True:
        time.sleep(1)  # cek setiap 1 detik apakah ada perubahan URL
        current_url = driver.current_url
        if current_url != last_url:
            # Jika URL berubah, simpan pasangan (dari -> ke)
            records.append([last_url, current_url])
            print(last_url, "->", current_url)
            last_url = current_url
except KeyboardInterrupt:
    # Program berhenti manual dengan CTRL + C
    print("\n🛑 Perekaman dihentikan oleh user.")
finally:
    # Simpan hasil rekaman navigasi ke file CSV
    with open("history_links.csv", "w", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        writer.writerow(["link_akses", "link_keluaran"])
        writer.writerows(records)

    # Tutup browser setelah selesai
    driver.quit()
    print("✅ Riwayat navigasi disimpan ke history_links.csv")
```
