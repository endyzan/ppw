ubah tiap url jadi simbol agar lebih mudah di pahami

```python
from itertools import product
import string

# Ambil semua kategori unik
unique_urls = filtered_df["url"].unique()

# Fungsi buat label a, b, ..., z, aa, ab, ...
def generate_labels(n):
    labels = []
    alphabet = string.ascii_lowercase
    for length in range(1, 10):  # panjang label 1, 2, dst.
        for combo in product(alphabet, repeat=length):
            labels.append(''.join(combo))
            if len(labels) == n:
                return labels
    return labels

# Generate label sebanyak jumlah kategori unik
labels = generate_labels(len(unique_urls))

# Buat mapping URL -> simbol
url_to_label = dict(zip(unique_urls, labels))

# Ganti isi kolom url dengan simbol
filtered_df["url"] = filtered_df["url"].map(url_to_label)

# Cek hasil
print(filtered_df.head())
```

### f. tiap sesi di buatkan waktu max 30 menit, jika lebih dari 30 menit berarti masuk ke sesi 2

Buat sesi berdasarkan timeout 30 menit

```python
import pandas as pd

df = pd.read_csv("session_host_time_url_NASA.csv", parse_dates=["datetime_readable"])

# Sortir dulu berdasarkan host lalu waktu
df = df.sort_values(by=["host", "datetime_readable"])

# Buat sesi kosong
df["REAL_SESSION"] = 0

current_session = 1
df.at[0, "REAL_SESSION"] = current_session

# Loop baris
for i in range(1, len(df)):
    same_host = df.iloc[i]["host"] == df.iloc[i-1]["host"]
    time_diff = (df.iloc[i]["datetime_readable"] - df.iloc[i-1]["datetime_readable"]).total_seconds() / 60
    
    # Jika host sama DAN jeda waktu <= 30 menit → sesi sama
    if same_host and time_diff <= 30:
        df.at[i, "REAL_SESSION"] = current_session
    else:
        current_session += 1
        df.at[i, "REAL_SESSION"] = current_session

df.to_csv("session_timeout_30_NASA.csv", index=False)
print("Sesi dengan timeout 30 menit berhasil dibuat.")
```
