# Tugas 9 | Crawling Twiter (X)

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

import time
import pandas as pd
```

```python
HASHTAG = "%23tesla"      # %23 = #
TARGET_TWEETS = 250
SCROLL_DELAY = 5
MAX_SCROLL = 120
```

```python
options = webdriver.ChromeOptions()
options.add_argument(r"--user-data-dir=C:\Users\endyzan\AppData\Local\Google\Chrome\User Data")
options.add_argument("--profile-directory=Default")
options.add_argument("--start-maximized")

driver = webdriver.Chrome(
    service=Service(ChromeDriverManager().install()),
    options=options
)
```

```python
url = f"https://x.com/search?q={HASHTAG}&f=live"
driver.get(url)

wait = WebDriverWait(driver, 30)
wait.until(EC.presence_of_element_located((By.XPATH, "//article")))
```

```python
tweets_data = set()
scroll_count = 0

while len(tweets_data) < TARGET_TWEETS and scroll_count < MAX_SCROLL:
    tweets = driver.find_elements(By.XPATH, "//article//div[@lang]")

    for tweet in tweets:
        text = tweet.text.strip()
        if text:
            tweets_data.add(text)

    print(f"Scroll {scroll_count+1} | Total tweet: {len(tweets_data)}")

    driver.find_element(By.TAG_NAME, "body").send_keys(Keys.END)
    time.sleep(SCROLL_DELAY)
    scroll_count += 1
```

```python
driver.quit()
```

```python
df = pd.DataFrame(list(tweets_data), columns=["tweet"])
df.to_csv("hashtag_250.csv", index=False, encoding="utf-8")

print(f"Selesai! Total tweet diambil: {len(df)}")
```
